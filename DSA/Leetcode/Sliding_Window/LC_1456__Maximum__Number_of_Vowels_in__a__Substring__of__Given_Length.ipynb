# Fixed Sliding Window – Maximum Number of Vowels in a Substring

# Pattern
Sliding Window (Fixed Window)

# Difficulty
Medium (Concept Building)

# Problem Statement
Given a string `s` and an integer `k`, return the maximum number of vowels present in any substring of length `k`.

# Key Idea
- The substring length remains fixed throughout the problem.
- Count the vowels in the first window of size `k`.
- Store it as the initial maximum vowel count.
- Slide the window one character at a time.
- If the character leaving the window is a vowel, decrease the vowel count.
- If the character entering the window is a vowel, increase the vowel count.
- Update the maximum vowel count after each slide.

```python
def maximumVowels(s,k):

  vowels = {"A","E","I","O","U","a","e","i","o","u"}

  vowel_count = 0

  for i in range(k):

    if s[i] in vowels:
      vowel_count+=1

  max_vowel_count = vowel_count

  for i in range(k,len(s)):

    if s[i-k]  in vowels:
      vowel_count-=1

    if s[i] in vowels:
      vowel_count+=1

    if vowel_count > max_vowel_count:
      max_vowel_count = vowel_count

  return max_vowel_count
```

```python
print(maximumVowels("abciiidef",3))
```

```python
print(maximumVowels("aeiou",2))
```

```python
print(maximumVowels("leetcode",3))
```

# Time Complexity: O(n)
# The first for loop counts the vowels in the initial window in O(k).
# The second for loop slides the window through the remaining characters in O(n - k).
# Since the loops are consecutive (not nested), the overall time complexity is:
# O(k) + O(n - k) = O(n)

# Space Complexity: O(1)
# A fixed-size set of vowels and a few integer variables are used.
# Their size does not depend on the input length, so the extra memory required is constant.
