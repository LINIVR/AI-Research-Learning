# Fixed Sliding Window – Maximum Average Subarray

# Pattern
Sliding Window (Fixed Window)

# Difficulty
Easy (Concept Building)

# Problem Statement
Find the maximum average that can be obtained from any contiguous subarray of length `k`.

# Key Idea
- Calculate the sum of the first window of size `k`.
- Store it as the initial maximum sum.
- Slide the window one position at a time.
- For each slide:
  - Remove the element leaving the window.
  - Add the element entering the window.
- Update the maximum sum if the current window sum is greater.
- Return the maximum average.

```python
def average_subarray(array,k):

  if not array or k<1 or k>len(array):
    return None

  current_sum =0

  for i in range(k):
    current_sum+=array[i]
  max_sum = current_sum

  for i in range(k,len(array)):
    current_sum+=array[i]-array[i-k]

    if current_sum>max_sum:
      max_sum =current_sum

  return  max_sum/k


```

```python
print(average_subarray([1,12,-5,-6,50,3], 4))
```

```python
print(average_subarray([5],1))
```

```python
print(average_subarray([1, 3, 2, 6, -1, 4, 1, 8, 2],5))
```

# Time Complexity: O(n)
# The first for loop calculates the sum of the initial window in O(k).
# The second for loop slides the window through the remaining elements in O(n - k).
# Since the loops are consecutive (not nested), the overall time complexity is:
# O(k) + O(n - k) = O(n)

# Space Complexity: O(1)
# Only two integer variables (current_sum and max_sum) are used.
# No additional data structure is created, so the extra memory required is constant.
