# Fixed Sliding Window – Maximum Points You Can Obtain from Cards

# Pattern
Sliding Window (Fixed Window)

# Difficulty
Medium

## Problem Statement
Given an integer array `cardPoints` and an integer `k`, return the maximum score that can be obtained by picking exactly `k` cards from either the beginning or the end of the array.

# Key Idea
- First, calculate the sum of the first `k` cards.
- Assume this is the initial maximum score.
- Gradually replace one card taken from the left with one card taken from the right.
- After every replacement, update the maximum score.
- Return the maximum score obtained.

```python
def maxScore(cardPoints,k):

    current_sum = 0

    for i in range(k):
      current_sum+=cardPoints[i]

    max_sum = current_sum

    for i in range(k):

      current_sum+=cardPoints[len(cardPoints)-i-1] -cardPoints[k-i-1]

      if current_sum>max_sum:
        max_sum = current_sum

    return max_sum
```

```python
# Example 1
cardPoints = [1, 2, 3, 4, 5, 6, 1]
k = 3
print("Example 1 Output:", maxScore(cardPoints, k))

# Example 2
cardPoints = [2, 2, 2]
k = 2
print("Example 2 Output:", maxScore(cardPoints, k))

# Example 3
cardPoints = [9, 7, 7, 9, 7, 7, 9]
k = 7
print("Example 3 Output:", maxScore(cardPoints, k))
```

# Time Complexity: O(k)
# The first loop calculates the sum of the first k cards.
# The second loop performs exactly k updates.
# Therefore, the total time complexity is O(k).

# Space Complexity: O(1)
# Only a few integer variables are used.
# No additional data structure is created.
