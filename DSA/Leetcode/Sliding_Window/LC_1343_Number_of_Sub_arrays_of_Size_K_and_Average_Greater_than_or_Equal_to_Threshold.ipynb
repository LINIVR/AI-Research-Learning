# Fixed Sliding Window – Number of Subarrays of Size K and Average Greater Than or Equal to Threshold

# Pattern
Sliding Window (Fixed Window)

# Difficulty
Medium

# Problem Statement
Given an integer array `arr`, an integer `k`, and an integer `threshold`, return the number of contiguous subarrays of size `k` whose average is greater than or equal to `threshold`.

# Key Idea
- Calculate the sum of the first window of size `k`.
- Instead of calculating the average every time, compare the window sum with `threshold × k`.
- Slide the window one position at a time.
- Update the window sum by removing the leaving element and adding the entering element.
- Count the number of valid subarrays.

```python
def numberSubarrays(arr,k,threshold):

  current_sum = 0
  count_subarray = 0
  target_sum = threshold*k

  for i in range(k):
    current_sum+=arr[i]

  if current_sum >= target_sum:
    count_subarray+=1

  for i in range(k,len(arr)):
    current_sum+=arr[i]-arr[i-k]

    if current_sum >= target_sum:
      count_subarray+=1

  return count_subarray
```

```python
print(numberSubarrays([2,2,2,2,5,5,5,8],3,4))
```

```python
print(numberSubarrays([11,13,17,23,29,31,7,5,2,3],3,5))
```

# Time Complexity: O(n)
# The first for loop calculates the sum of the initial window in O(k).
# The second for loop slides the window through the remaining elements in O(n - k).
# Since the loops are consecutive (not nested),
# the overall time complexity is O(k) + O(n - k) = O(n).

# Space Complexity: O(1)
# Only a few integer variables are used.
# No additional data structure is created.
