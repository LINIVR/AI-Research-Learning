# Variable Sliding Window – Longest Substring Without Repeating Characters

# Pattern
Variable Sliding Window

# Difficulty
Medium

# Problem Statement
Given a string `s`, return the length of the longest substring without repeating characters.

# Key Idea
- Expand the window using the right pointer.
- If a duplicate character is found, shrink the window from the left until the duplicate is removed.
- Keep track of the maximum window length.

```python
def longestSubString(s):

    max_length = 0
    window = set()
    left =0
    for right in range(len(s)):

      while s[right] in window:
        window.remove(s[left])
        left+=1

      window.add(s[right])
      current_length = right-left+1

      if current_length>max_length:
        max_length = current_length


    return max_length


```

```python

# Example 1
s = "abcabcbb"
print("Example 1")
print("Input :", s)
print("Output:", longestSubString(s))
print("Expected Output: 3")
print()

# Example 2
s = "bbbbb"
print("Example 2")
print("Input :", s)
print("Output:", longestSubString(s))
print("Expected Output: 1")
print()

# Example 3
s = "pwwkew"
print("Example 3")
print("Input :", s)
print("Output:", longestSubString(s))
print("Expected Output: 3")
```

### Problem

Find the length of the longest substring without repeating characters.

### Pattern

Variable Sliding Window

### Variables

- `window` stores the unique characters in the current window.
- `left` represents the beginning of the window.
- `right` expands the window.
- `max_length` stores the maximum window length.

### Algorithm

1. Create an empty set to store unique characters.
2. Initialize `left` and `max_length` to 0.
3. Move the `right` pointer through the string.
4. If the current character already exists in the window, remove characters from the left until the duplicate is removed.
5. Add the current character to the window.
6. Calculate the current window length.
7. Update the maximum length.
8. Return the maximum length.

### Return

Return the length of the longest substring without repeating characters.

# Worst-Case Time Complexity: O(n)
# Explanation:
# - The right pointer traverses the string from left to right only once.
# - The left pointer also traverses the string from left to right only once.
# - Each character is added to the set at most one time.
# - Each character is removed from the set at most one time.
# Therefore, the total number of operations is proportional to n.

# Worst-Case Space Complexity: O(n)
# Explanation:
# - In the worst case, the string contains all unique characters.
# - The set stores every character in the current window.
# - Therefore, the size of the set can grow up to n characters.
# Hence, the worst-case space complexity is O(n).
