# LeetCode 26 – Remove Duplicates from Sorted Array

# Pattern
Two Pointers

# Difficulty
Easy

# Key Idea
- The array is already sorted.
- Use two pointers:
  - `left` points to the last unique element.
  - `right` scans the array to find the next unique element.
- If `nums[right]` is different from `nums[left]`, move `left` one step forward and overwrite that position with `nums[right]`.
- The first `k` positions of the array contain the unique elements.
- Return `k`, the number of unique elements.

```python
def removeDuplicates(nums):

  left = 0
  right = 1

  while right <len(nums):

    if nums[left] != nums[right]:

      left+=1

      nums[left] = nums[right]


    right+=1

  return left+1

```

```python
nums = [1,2,2,2,6,7]

k = removeDuplicates(nums)

print("No: of unique elements:", k)

print("Modified Array:", nums)

print("Unique elements:",nums[:k])
```

```python
nums = [0,0,1,1,1,2,2,3,3,3,3,4]

k = removeDuplicates(nums)

print("No: of unique elements:", k)

print("Modified Array:", nums)

print("Unique elements:",nums[:k])

```

# Time Complexity: O(n)
# The right pointer traverses the array only once.
# Each element is processed at most once, so the total time required is linear.

# Space Complexity: O(1)
# The algorithm modifies the array in place without creating any additional data structure.
# Only the pointer variables (left and right) are used, which require a constant amount of extra memory.
