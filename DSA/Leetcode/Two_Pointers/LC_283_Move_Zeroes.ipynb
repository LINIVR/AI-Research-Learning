# LeetCode 283 – Move Zeroes

# Pattern
Two Pointers

# Difficulty
Easy

# Key Idea
- Use two pointers:
  - `right` is the scanning pointer that traverses the entire array.
  - `left` is the placement pointer that indicates where the next non-zero element should be placed.
- Whenever `nums[right]` is non-zero, swap `nums[left]` and `nums[right]`.
- Increment both pointers after a successful swap.
- If `nums[right]` is zero, increment only `right`.
- The algorithm modifies the array in place while maintaining the relative order of the non-zero elements.

```python
def move_zeroes(nums):

  left = 0
  right = 0

  while right < len(nums):

    if nums[right] != 0 :

      nums[left],nums[right] = nums[right],nums[left]

      left+=1

    right+=1

  return nums
```

```python

print(move_zeroes([0, 1, 0, 3, 12]))

print(move_zeroes([2, 5, 7, 0, 0, 9, 0, 4]))
```

# Time Complexity: O(n)
# The right pointer traverses the array only once, and each element is processed at most once.
# Therefore, the total time required is linear.

# Space Complexity: O(1)
# The algorithm modifies the array in place without creating any additional data structure.
# Only the pointer variables (left and right) are used, which require a constant amount of extra memory.
