# LeetCode 344 – Reverse String

# Pattern
Two Pointers

# Difficulty
Easy

# Key Idea
- Initialize two pointers.
- `left` starts from the beginning.
- `right` starts from the end.
- Swap the characters at `left` and `right`.
- Move `left` forward and `right` backward.
- Continue until the pointers meet.

```python
def reverse_string(s):

  left = 0
  right = len(s)-1

  while left < right:

    s[left],s[right] = s[right],s[left]

    left+=1
    right-=1

  return s

```

```python
print(reverse_string(['h','e','l','l','o']))
print(reverse_string(["H","a","n","n","a","h"]))
```

# Time Complexity: O(n)
# The left and right pointers move toward the center, and each element is visited at most once.
# Therefore, the total time required is linear.

# Space Complexity: O(1)
# The string is modified in place without using any additional data structure.
# Only two pointer variables are used, so the extra space required is constant.
