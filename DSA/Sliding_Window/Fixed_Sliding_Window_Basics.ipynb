# Fixed Sliding Window – Basics

## Pattern
Sliding Window (Fixed Window)

## Key Idea
- The window size remains fixed throughout the algorithm.
- First, calculate the sum of the initial window of size `k`.
- Then slide the window one position at a time.
- For each slide:
  - Remove the element leaving the window.
  - Add the element entering the window.
- Update the maximum sum after each slide.

## Example

Array = [2, 1, 5, 1, 3, 2]

Window Size = 3

Window 1: [2, 1, 5] → Sum = 8

Window 2: [1, 5, 1] → Sum = 7

Window 3: [5, 1, 3] → Sum = 9

Window 4: [1, 3, 2] → Sum = 6

Maximum Sum = 9

```python
def largest(array,k=3):

  current_sum = 0
  for i in range(k):
    current_sum+=array[i]
  max_sum = current_sum

  for i in range(k,len(array)):
    current_sum+=array[i]-array[i-k]

    if current_sum>max_sum:
      max_sum=current_sum

  return max_sum
```

```python
print(largest([2, 1, 5, 1, 3, 2]))
```

```python
print(largest([1,2,3,4,5]))
```

# Time Complexity: O(n)
# The first for loop calculates the sum of the initial window of size k.
# The second for loop slides the window through the array one position at a time.
# Since the two loops are consecutive (not nested), the total time complexity is:
# O(k) + O(n - k) = O(n)

# Space Complexity: O(1)
# Only a constant amount of extra memory is used for variables such as
# current_sum and max_sum.
# No additional array or data structure is created.

# Note:
# Consecutive loops -> Add the time complexities.
# O(k) + O(n - k) = O(n)
#
# Nested loops -> Multiply the time complexities.
# O(n) × O(n) = O(n²)
