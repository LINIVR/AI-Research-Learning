# LeetCode 167 – Two Sum II: Input Array Is Sorted

# Pattern
Two Pointers

# Difficulty
Medium

# Key Idea
- The array is already sorted.
- Initialize `left` at the beginning and `right` at the end.
- Calculate the current sum.
- If the sum is equal to the target, return the 1-based indices.
- If the sum is less than the target, move `left` to increase the sum.
- If the sum is greater than the target, move `right` to decrease the sum.

```python
def two_sum(numbers,target):
  left = 0
  right = len(numbers)-1


  while left < right:

    current_sum = numbers[left]+numbers[right]

    if current_sum == target:
      return [left+1,right+1]

    elif current_sum<target:
      left+=1

    else:
      right-=1

  return None
```

```python
print(two_sum([2,7,11,15],9))

print(two_sum([2,3,4],6))

print(two_sum([-1,0], -1))
```

# Time Complexity: O(n)
# The left and right pointers traverse the array at most once.
# Therefore, the total time required is linear.

# Space Complexity: O(1)
# Only two pointer variables and one integer variable (current_sum) are used.
# No additional array or data structure is created.
# Therefore, the extra space required is constant.
